```python
import requests
```

```python
url = "https://api.v2.urbanobservatory.ac.uk/sensors/data/csv"

response = requests.get(url)

if response.status_code == 200:
    # Save to file
    with open("urban_observatory_data.csv", "wb") as f:
        f.write(response.content)
    print("CSV data saved successfully.")
else:
    print(f"Failed to fetch data. Status code: {response.status_code} message {response.text}")
```

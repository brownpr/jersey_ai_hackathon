```python
import requests
```

```python
URL = "http://localhost:8001/predict"
```

```python
payload = {
    "samples": [
        {
            "hour": 0,
            "day": 1,
            "weekday": 0,
            "month": 9,
            "year": 2025,
            "plates_in": 7.5,
            "plates_matching": 2.0,
            "plates_out": 8.5,
            "is_rush_hour": 0,
            "is_weekend": 0
        },
        {
            "hour": 0,
            "day": 1,
            "weekday": 5,
            "month": 11,
            "year": 2025,
            "plates_in": 17.433333,
            "plates_matching": 5.833333,
            "plates_out": 16.8,
            "is_rush_hour": 0,
            "is_weekend": 1
        }
    ]
}
```

```python
def main():
    try:
        response = requests.post(URL, json=payload)
        print(f"Status code: {response.status_code}")
        print("Response JSON:")
        print(response.json())
    except Exception as e:
        print(f"Error calling API: {e}")
```

```python
main()
```
